## 📌 Justificativa Técnica
Dados de sensores de perfuração são ruidosos e sujeitos a falhas eletrônicas. Para garantir que o modelo aprenda a geologia real e não os artefatos do sensor, aplicamos um pipeline de limpeza focado em:
1. **Sanitização de Range:** GR limitado a 0-250 API.
2. **Interpolação Temporal:** Para garantir continuidade nos logs.
3. **Sinais Derivados:** Gradientes e médias móveis para capturar tendências de formação.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Carregando os dados processados
df_cleaned = pd.read_parquet("../data/processed/train_cleaned.parquet")

print(f"Volume total de registros processados: {len(df_cleaned):,}")
print(f"Total de poços únicos: {df_cleaned['well_id'].nunique()}")
```

## 📊 Comparativo de Sinais: Original vs Suavizado
Visualizando o efeito da média móvel no Gamma Ray.

```python
sample_well = df_cleaned[df_cleaned['well_id'] == df_cleaned['well_id'].unique()[0]]

plt.figure(figsize=(15, 6))
plt.plot(sample_well['MD'], sample_well['GR'], label='Original (Clipped)', color='gray', alpha=0.5)
plt.plot(sample_well['MD'], sample_well['GR_rolling_mean'], label='Suavizado (Rolling Mean)', color='#2A9D8F', linewidth=2)
plt.title("Impacto da Suavização no Log de Gamma Ray", fontsize=14)
plt.xlabel("Depth (MD)")
plt.ylabel("GR API")
plt.legend()
plt.show()
```

## 🔎 Engenharia de Atributos: Gradiente de GR
O gradiente (`GR_diff`) ajuda o modelo a detectar 'quebras' geológicas.

```python
plt.figure(figsize=(15, 4))
plt.plot(sample_well['MD'], sample_well['GR_diff'], color='#E76F51')
plt.title("Gradiente de Gamma Ray (GR_diff)", fontsize=14)
plt.xlabel("Depth (MD)")
plt.ylabel("Delta GR")
plt.show()
```

## ✅ Conclusão da Auditoria
A base está agora pronta para a etapa de modelagem. A redução de ruído e a criação de atributos diferenciais reduzem o risco de overfitting em anomalias de sensores.
