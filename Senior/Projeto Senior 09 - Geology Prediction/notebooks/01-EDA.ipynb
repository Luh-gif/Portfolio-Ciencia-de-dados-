## 📌 Objetivo de Negócio
O objetivo desta análise é explorar o dataset geológico para identificar padrões de resposta geofísica (Gamma Ray) e relações espaciais que permitam prever a formação geológica em poços horizontais. A automação deste processo reduz o tempo de interpretação e aumenta a precisão da perfuração direcional (Geosteering).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os

# Configurações Premium de Visualização
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("viridis")
COLOR_PRIMARY = "#2A9D8F"
COLOR_SECONDARY = "#264653"
```

## 📂 Carregamento e Inventário de Dados

```python
RAW_DATA_PATH = "../data/raw/train/"
horizontal_wells = glob.glob(os.path.join(RAW_DATA_PATH, "*__horizontal_well.csv"))
type_wells = glob.glob(os.path.join(RAW_DATA_PATH, "*__typewell.csv"))

print(f"Total de Poços Horizontais: {len(horizontal_wells)}")
print(f"Total de Poços de Referência (Typewells): {len(type_wells)}")
```

## 📊 Análise de Estrutura de um Poço Amostra

```python
# Lendo um poço aleatório para inspeção
sample_h_well = pd.read_csv(horizontal_wells[0])
sample_t_well = pd.read_csv(type_wells[0])

print("--- Estrutura Poço Horizontal ---")
print(sample_h_well.info())

print("\n--- Estrutura Typewell ---")
print(sample_t_well.info())
```

## 📈 Distribuição de Gamma Ray (GR)
O GR é o principal indicador de litologia (argilosidade).

```python
plt.figure(figsize=(12, 5))
sns.histplot(sample_h_well['GR'], kde=True, color=COLOR_PRIMARY)
plt.title("Distribuição de Gamma Ray - Poço Amostra", fontsize=14, pad=15)
plt.xlabel("Gamma Ray (API)")
plt.ylabel("Frequência")
plt.show()
```

## 📍 Visualização Espacial (MD vs GR)
Analisando como a geologia varia ao longo da profundidade medida (MD).

```python
plt.figure(figsize=(15, 4))
plt.plot(sample_h_well['MD'], sample_h_well['GR'], color=COLOR_SECONDARY, linewidth=1)
plt.fill_between(sample_h_well['MD'], sample_h_well['GR'], color=COLOR_PRIMARY, alpha=0.3)
plt.title("Log de Gamma Ray ao longo do Poço Horizontal", loc='left', fontsize=14)
plt.xlabel("Measured Depth (MD)")
plt.ylabel("GR")
plt.tight_layout()
plt.show()
```

## 🔍 Verificação de Missings e Outliers

```python
missing_data = sample_h_well.isnull().sum() / len(sample_h_well) * 100
print("Porcentagem de Dados Ausentes:")
print(missing_data[missing_data > 0])
```
